```python
import numpy as np
```

**1. Advanced Scalar &amp; Array Operations**

```python

scalar = 4.3

array = np.array([1.2, -2.5, 3.7, 4.1, -5.3, 6.8, 7.4, -8.2])

print("Scalar:", scalar)
print("Array:", array)
```

```python

print("Add:", array + scalar)
print("Sub:", array - scalar)
print("Mul:", array * scalar)
print("Div:", array / scalar)
```

```python
print("Abs Value:",np.abs(array))
print("Square Root:",np.sqrt(np.abs(array)))
print("Exponential:",np.exp(array))
print("Natural Log:",np.log(np.abs(array)))
print("Log Base 10:",np.log10(np.abs(array)))
```

```python
print("Sin: ",np.sin(array))
print("cos: ",np.cos(array))
print("tan: ",np.tan(array))
```

```python
clipped_array = np.clip(array, -1, 1)

print("Clipped Array:",clipped_array)
print("\narcsin:",np.arcsin(clipped_array))
print("\narccos:",np.arccos(clipped_array))
print("\narctan:",np.arctan(array))
```

**2. Matrix Mathematical Operations**

```python
matrix = np.array([
    [1.5, -2.3, 3.7],
    [-4.2, 5.6, -6.1],
    [7.4, -8.5, 9.2],
    [-3.3, 2.8, -1.7]
])

print("4x3 Matrix:",matrix)
```

```python
print("sin:",np.sin(matrix))
print("cos:",np.cos(matrix))
print("\ntan:",np.tan(matrix))
```

```python
clipped_matrix = np.clip(matrix, -1, 1)

print("arcsin:",np.arcsin(clipped_matrix))
print("\narccos:",np.arccos(clipped_matrix))
print("\narctan:",np.arctan(matrix))
```

```python
print("Exponential:",np.exp(matrix))
print("Natural Log:",np.log(np.abs(matrix)))
print("Log Base 10:",np.log10(np.abs(matrix)))
```

```python
print("Abs Value:",np.abs(matrix))
print("Square Root:",np.sqrt(np.abs(matrix)))
print("Square:",np.square(matrix))
print("Cube:",np.cbrt(matrix))
```

```python
print("Remainder when divided by 3:",np.mod(matrix, 3))
print("Round:",np.round(matrix))
print("Floor:",np.floor(matrix))
print("Ceil:",np.ceil(matrix))
```

**3. Matrix Statistics and Analysis**

```python
random_matrix= np.random.uniform(-10, 10, (3, 4))
print("Random 3x4 Matrix:",random_matrix)
```

```python
print("Maximum:", np.max(random_matrix))
print("Minimum:", np.min(random_matrix))
print("Index of Maximum:", np.unravel_index(np.argmax(random_matrix), random_matrix.shape))
print("Index of Minimum:", np.unravel_index(np.argmin(random_matrix), random_matrix.shape))
print("Sum:", np.sum(random_matrix))
print("Product:", np.prod(random_matrix))
print("Mean:", np.mean(random_matrix))
print("Median:", np.median(random_matrix))
print("Variance:", np.var(random_matrix))
print("Standard Deviation:", np.std(random_matrix))
```
