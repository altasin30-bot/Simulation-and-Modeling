**Task 1: Scalar & Array Operations**

```python
import numpy as np
```

```python
scalar = 3.7
print("Scalar =", scalar)
```

```python
arr = np.array([1.2, 2.5, 3.7, 4.1, 5.8, 6.3])

print("Array =", arr)
```

```python
print("sin =", np.sin(arr))
print("cos =", np.cos(arr))
print("tan =", np.tan(arr))

print("Exponential =", np.exp(arr))
print("Logarithm =", np.log(arr))

print("Absolute =", np.abs(arr))
print("Square Root =", np.sqrt(arr))
```

**Task 2: 3×2 Matrix Operations**

```python
matrix = np.array([
    [0.5, 1.2],
    [2.0, 2.5],
    [0.8, 1.5]
])

print("3x2 Matrix:")
print(matrix)
```

```python
print("Sin:")
print(np.sin(matrix))

print("Cos:")
print(np.cos(matrix))

print("Tan:")
print(np.tan(matrix))
```

```python
print("Exponential:")
print(np.exp(matrix))
```

```python
print("Natural Log:")
print(np.log(matrix))

print("Log base 10:")
print(np.log10(matrix))
```

```python
print("Absolute Value:")
print(np.abs(matrix))
```

```python
print("Square Root:")
print(np.sqrt(matrix))
```

```python
print("Remainder when divided by 2:")
print(np.remainder(matrix, 2))
```

```python
print("Round:")
print(np.round(matrix))

print("Floor:")
print(np.floor(matrix))

print("Ceil:")
print(np.ceil(matrix))
```

```python
safe_matrix = np.clip(matrix, -1, 1)

print("Arcsin:")
print(np.arcsin(safe_matrix))

print("Arccos:")
print(np.arccos(safe_matrix))
```

**Task 4: Vector Functions**

```python
matrix2 = np.array([
    [2.1, 4.7, 1.5],
    [3.8, 6.2, 2.9]
])

print("2x3 Matrix:")
print(matrix2)
```

```python
print("Maximum value:", np.max(matrix2))
print("Minimum value:", np.min(matrix2))
print("Maximum index:", np.argmax(matrix2))
print("Minimum index:", np.argmin(matrix2))
print("Vector length:", np.linalg.norm(matrix2))
```

```python
sorted_rows = np.sort(matrix2, axis=1)

print("Row-wise ascending:")
print(sorted_rows)
print("Row-wise descending:")
print(np.sort(matrix2, axis=1)[:, ::-1])
```

```python
flattened = matrix2.flatten()

print("Flattened:")
print(flattened)
print("Entire matrix descending:")
print(np.sort(flattened)[::-1])
```

```python
print("Sum:", np.sum(matrix2))
print("Product:", np.prod(matrix2))
print("Median:", np.median(matrix2))
print("Mean:", np.mean(matrix2))
print("Standard Deviation:", np.std(matrix2))
```
